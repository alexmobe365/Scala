```scala
def maxHand(mano1: Int, mano2: Int): Int = {
  if (mano1 > mano2) {
    mano1
  } else {
    mano2
  }
}

println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (bust(handA) && bust(handB)) {
    0
  } else if (bust(handA)) {
    handB
  } else if (bust(handB)) {
    handA
  } else if (handA > handB) {
    handA
  } else {
    handB
  }
}
println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

```scala
def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

def mejorMano(manos: Array[Int]): Int = {
  var mejor = 0
  var i = 0

  while (i < manos.length) {
    if (!bust(manos(i)) && manos(i) > mejor) {
      mejor = manos(i)
    }
    i += 1
  }

  mejor
}


// Ronda 1

val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

println("=== RONDA 1 ===")

jugadores.zip(manos).foreach { caso =>
  val jugador = caso._1
  val mano = caso._2

  if (bust(mano)) {
    println(s"$jugador: $mano → Se ha pasado")
  } else {
    println(s"$jugador: $mano → Sigue")
  }
}

println(s"Mejor puntuación válida: ${mejorMano(manos)}")


// Ronda 2

val manosRonda2 = Array(22, 19, 20, 21)

println("=== RONDA 2 ===")

jugadores.zip(manosRonda2).foreach { caso =>
  val jugador = caso._1
  val mano = caso._2

  if (bust(mano)) {
    println(s"$jugador: $mano → Se ha pasado")
  } else {
    println(s"$jugador: $mano → Sigue")
  }
}

println(s"Mejor puntuación válida: ${mejorMano(manosRonda2)}")
```
